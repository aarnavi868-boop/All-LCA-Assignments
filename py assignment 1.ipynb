```python
# Assignment 1
# Operations on Dictionary, Tuple and List

# List of students
students_list=['Aarav','Riya','Vivaan']

# Tuple of students
students_tuple=('Aarav','Riya','Vivaan')

# Dictionaries of students
students={1:'Aarav', 2:'Riya', 3:'Vivaan'}

print("Students list:",students_list)
print("Students tuple:",students_tuple)
print("Students dictionary:",students)

#Add a student
students[4]='Ananya'
print('\nAfter Adding a Student:')
print(students)

#Update a Student
students[2]='Ishita'
print('\nAfter Updating a Student:')
print(students)

#Delete a Student
del students[3]
print('\nAfter Deleting a Student:')
print(students)
```
